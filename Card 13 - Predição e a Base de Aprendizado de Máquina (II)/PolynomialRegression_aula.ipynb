## Polynomial Regression

E se os seus dados não parecerem nada lineares? Vamos olhar para alguns dados de velocidade da página / valor de compra com uma aparência mais realista:

```python
import numpy as np
import matplotlib.pyplot as plt #bibliotecas usadas para regressao poli e fazer os graficos

np.random.seed(2) 
pageSpeeds = np.random.normal(3.0, 1.0, 1000) #desvio padrao de x
purchaseAmount = np.random.normal(50.0, 10.0, 1000) / pageSpeeds #formula para y

plt.scatter(pageSpeeds, purchaseAmount) 
```

O numpy tem uma função polyfit bem útil, que nos permite construir um modelo polinomial de grau n dos nossos dados que minimiza o erro quadrático. Vamos testar com um polinômio de 4º grau:

```python
x = np.array(pageSpeeds) #colocando os valores de paginas em um array
y = np.array(purchaseAmount) # mesma coisa para valor de compra

p4 = np.poly1d(np.polyfit(x, y, 4)) #criando o polinomio de 
```

Vamos visualizar nosso gráfico de dispersão original, junto com um gráfico dos valores previstos usando o polinômio para tempos de velocidade da página variando de 0 a 7 segundos:

```python
import matplotlib.pyplot as plt

xp = np.linspace(0, 7, 100)
plt.scatter(x, y)
plt.plot(xp, p4(xp), c='r')
plt.show()
```

Parece bem bom! Vamos medir o erro R-quadrado:

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, p4(x))

print(r2)
```

## Atividade

Tente diferentes ordens polinomiais. Você consegue um ajuste melhor com ordens mais altas? Você começa a ver overfitting, mesmo que o valor de R-quadrado pareça bom para esse conjunto de dados específico?

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

p15 = np.poly1d(np.polyfit(x, y, 15))#aumentei o grau de 4 para 15

xp = np.linspace(0, 7, 100)

plt.scatter(x, y)
plt.plot(xp, p15(xp), c='r')
plt.show()

r2 = r2_score(y, p15(x))
print(r2) # com o grau em 15, valor de r2 ficou bom, porem no grafico por ter um grau muito alto tem picos e vales agressivos evidenciando um overfitting
```
