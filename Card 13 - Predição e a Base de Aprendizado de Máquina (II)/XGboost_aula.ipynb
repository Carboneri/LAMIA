Vamos experimentar usando o dataset Iris. Esse conjunto de dados inclui a largura e o comprimento das pétalas e sépalas de várias flores Iris, e a espécie específica de Iris à qual a flor pertence. Nosso desafio é prever a espécie de uma amostra de flor apenas com base no tamanho de suas pétalas. Vamos revisitar esse conjunto de dados mais adiante também, quando falarmos sobre análise de componentes principais.

```python
from sklearn.datasets import load_iris #importa o dataset iris

iris = load_iris() #variavel para receber o dataset

numSamples, numFeatures = iris.data.shape #extrai os dados do dataset
print(numSamples)
print(numFeatures)
print(list(iris.target_names))
```

Vamos dividir nossos dados em 20% reservados para testar nosso modelo, e os 80% restantes para treiná-lo. Ao reservar nossos dados de teste, podemos garantir que estamos avaliando os resultados com base em flores novas que o modelo nunca viu. Normalmente nos referimos às nossas features (nesse caso, os tamanhos das pétalas) como X, e aos rótulos (nesse caso, as espécies) como y.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=0) #dividi os dados em duas partes: dados de treinamento e dados de teste
```

Agora vamos carregar o XGBoost e converter nossos dados para o formato DMatrix que ele espera. Um para os dados de treino, e um para os dados de teste.

```python
import xgboost as xgb
#transforma os dados separados anteriormente X_train e X_test nesse formato específico.
train = xgb.DMatrix(X_train, label=y_train)
test = xgb.DMatrix(X_test, label=y_test)
```

Agora vamos definir nossos hiperparâmetros. Estamos escolhendo softmax já que esse é um problema de classificação múltipla, mas os outros parâmetros idealmente deveriam ser ajustados por meio de experimentação.

```python
param = {
    'max_depth': 4,
    'eta': 0.3,
    'objective': 'multi:softmax',
    'num_class': 3}
epochs = 10 #define as configuracoes que sao os hiperparametros que o algortimo vai usar para aprender
```

Vamos treinar nosso modelo usando esses parâmetros como um primeiro palpite.

```python
model = xgb.train(param, train, epochs) #aqui acontece o aprendizado
```

Agora vamos usar o modelo treinado para prever as classificações dos dados que reservamos para teste. Cada número de classificação que obtemos corresponde a uma espécie específica de Iris.

```python
predictions = model.predict(test) #manda o modelo ja treinado "adivinhar" as especies das flores da base de teste
```

```python
print(predictions)
```

Vamos medir a acurácia nos dados de teste...

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, predictions) #mede se o modelo foi bem
```
