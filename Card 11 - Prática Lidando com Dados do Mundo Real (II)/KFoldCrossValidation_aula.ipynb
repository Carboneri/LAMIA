```python
import numpy as np
from sklearn import model_selection 
from sklearn import datasets
from sklearn import svm

iris = datasets.load_iris()  # carrega o dataset iris, classico de classificacao
```

```python
# usando o model_selection p dividir os dados em treino e teste
X_train, X_test, y_train, y_test = model_selection.train_test_split(
    iris.data, iris.target, test_size=0.4, random_state=0
)  # 60% treino, 40% teste, random_state fixa a divisao (sempre a msm)

# construindo o modelo
clf = svm.SVC(kernel='linear', C=1).fit(X_train, y_train)  # treina um SVM c kernel linear

# medindo a performance
clf.score(X_test, y_test)  # acuracia do modelo em cima dos dados de teste
```

```python
# usando o model_selection p calcular os scores com k-fold cross validation
scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv=5)  # cv=5 = divide em 5 folds

# CORRECAO: colocado os parenteses no print pro Python 3
print(scores)  # acuracia de cada um dos 5 folds

# media dos 5 folds, da uma nocao mais confiavel do desempenho real do modelo
print(scores.mean())
```

```python
# construindo o modelo com um kernel diferente ('poly') para comparar
clf = svm.SVC(kernel='poly', C=1).fit(X_train, y_train)

# usando model_selection (atualizado) no lugar do cross_validation (antigo, desatualizado)
scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv=5)

print(scores)
print(scores.mean())
```

```python
# treina de novo o SVC c kernel poly, igualzinho a celula de cima
clf = svm.SVC(kernel='poly', C=1).fit(X_train, y_train)

# agora mede só no split fixo de teste (sem cross validation), da um numero dif do score medio do cv
clf.score(X_test, y_test)
```
