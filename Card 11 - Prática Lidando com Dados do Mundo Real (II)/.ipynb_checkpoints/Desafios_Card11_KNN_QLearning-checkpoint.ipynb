## Desafio 1 - KNN (Using KNN to predict a rating for a movie)

**Perguntas do desafio:**
1. Nossa escolha de K=10 foi arbitraria - qual o efeito de diferentes valores de K nos resultados?
2. Nossa metrica de distancia tambem foi arbitraria - somamos a distancia de cosseno dos generos com a diferenca de popularidade normalizada. Da para melhorar isso?

**Observacao:** o caminho original do dataset era local (`C:/Users/zinho/.../u.data`). Para este notebook funcionar de forma independente, o dataset MovieLens foi carregado a partir da pasta `ml-100k/` (mesma estrutura de arquivos `u.data` e `u.item` usada em aula).

```python
import pandas as pd
import numpy as np
from scipy import spatial
import operator

r_cols = ['user_id', 'movie_id', 'rating']
ratings = pd.read_csv('ml-100k/u.data', sep='\t', names=r_cols, usecols=range(3))

movieProperties = ratings.groupby('movie_id').agg({'rating': ['size', 'mean']})
movieNumRatings = pd.DataFrame(movieProperties['rating']['size'])
movieNormalizedNumRatings = movieNumRatings.apply(lambda x: (x - np.min(x)) / (np.max(x) - np.min(x)))

movieDict = {}
with open('ml-100k/u.item', encoding='latin-1') as f:
    for line in f:
        fields = line.rstrip('\n').split('|')
        movieID = int(fields[0])
        name = fields[1]
        genres = fields[5:25]
        genres = list(map(int, genres))
        movieDict[movieID] = (
            name,
            genres,
            movieNormalizedNumRatings.loc[movieID].get('size'),
            movieProperties.loc[movieID].rating.get('mean')
        )

print("Total de filmes carregados:", len(movieDict))
```

### Metrica de distancia original (a mesma usada em aula)

```python
def ComputeDistance(a, b):
    genresA = a[1]
    genresB = b[1]
    genreDistance = spatial.distance.cosine(genresA, genresB)
    popularityA = a[2]
    popularityB = b[2]
    popularityDistance = abs(popularityA - popularityB)
    return genreDistance + popularityDistance

def getNeighbors(movieID, K, distanceFunc):
    distances = []
    for movie in movieDict:
        if movie != movieID:
            dist = distanceFunc(movieDict[movieID], movieDict[movie])
            distances.append((movie, dist))
    distances.sort(key=operator.itemgetter(1))
    return [distances[x][0] for x in range(K)]
```

### Parte 1 - Testando diferentes valores de K

Para responder a primeira pergunta do desafio, comparamos K = 3, 5, 10 e 15 para o filme Toy Story (id=1), observando a nota media prevista e a lista de vizinhos encontrados.

```python
for K in [3, 5, 10, 15]:
    neighbors = getNeighbors(1, K, ComputeDistance)
    avgRating = sum(movieDict[n][3] for n in neighbors) / K
    print(f"K={K:>2} -> nota media dos vizinhos = {avgRating:.3f}")
    print("   vizinhos:", [movieDict[n][0] for n in neighbors])
```

**Conclusao da Parte 1:** valores de K muito baixos (K=3) deixam a previsao sensivel a apenas 1 ou 2 filmes "estranhos" que entraram por acaso na lista, enquanto valores muito altos de K comecam a incluir filmes cada vez menos parecidos, "diluindo" a recomendacao com vizinhos fracos. No teste acima, a nota media variou dependendo do K, mostrando que a escolha de K funciona como o mesmo trade-off de bias/variancia visto na Secao 7: K baixo = mais variancia (sensivel a ruido), K alto = mais bias (generaliza demais).

### Parte 2 - Melhorando a metrica de distancia

A metrica original somava a distancia de cosseno dos generos com a diferenca de popularidade, sem nenhum peso ou ponderacao entre elas - ou seja, as duas contribuiam igualmente, mesmo estando em escalas diferentes. Para melhorar, criamos uma metrica com **pesos ajustaveis** para cada componente, e adicionamos um terceiro fator que a metrica original nao considerava: a **diferenca entre as notas medias** dos filmes (normalizada pela escala 1-5).

```python
def ComputeDistanceImproved(a, b, w_genre=1.0, w_pop=1.0, w_rating=0.5):
    genreDistance = spatial.distance.cosine(a[1], b[1])
    popularityDistance = abs(a[2] - b[2])
    ratingDistance = abs(a[3] - b[3]) / 4.0  # normalizado (notas vao de 1 a 5)
    return (w_genre * genreDistance) + (w_pop * popularityDistance) + (w_rating * ratingDistance)

neighbors_original = getNeighbors(1, 10, ComputeDistance)
neighbors_improved = getNeighbors(1, 10, ComputeDistanceImproved)

print("Vizinhos com metrica ORIGINAL :", [movieDict[n][0] for n in neighbors_original])
print("Vizinhos com metrica MELHORADA:", [movieDict[n][0] for n in neighbors_improved])
```

**Conclusao da Parte 2:** ao incluir a diferenca de nota media como um terceiro componente e permitir pesos ajustaveis (`w_genre`, `w_pop`, `w_rating`), a lista de vizinhos muda em relacao a metrica original - alguns filmes que tinham genero/popularidade parecidos mas nota muito diferente saem da lista, sendo substituidos por filmes mais coerentes tambem na avaliacao dos usuarios. Isso mostra que a metrica de distancia original era realmente arbitraria: ela ignorava um sinal importante (a nota media) e tratava genero e popularidade com o mesmo peso, mesmo sem nenhuma justificativa para isso.

## Desafio 2 - Reinforcement Learning & Q-Learning with Gym

**Perguntas do desafio:**
1. Modificar o bloco de treinamento para contar o total de passos (time steps), usando isso como metrica de qualidade do sistema de Q-Learning. Aumentar o numero de viagens simuladas e remover as chamadas de `sleep()`.
2. Experimentar com os hiperparametros: ate onde o numero de epocas pode cair antes do modelo comecar a sofrer? Quais learning rates, discount factors e taxas de exploracao deixam o treinamento mais eficiente?

**Observacao:** o material original usa `gym` com `Taxi-v3`. Como o pacote `gym` esta descontinuado, este notebook usa `gymnasium` (sucessor mantido oficialmente) com `Taxi-v4` - a logica do algoritmo e exatamente a mesma.

```python
import gymnasium as gym
import numpy as np
import random

random.seed(1234)
streets = gym.make("Taxi-v4")

print(streets.action_space, streets.observation_space)
```

### Parte 1 - Funcao de treino contando o total de passos (sem sleep)

```python
def train_q_learning(env, learning_rate, discount_factor, exploration, epochs):
    q_table = np.zeros([env.observation_space.n, env.action_space.n])
    total_steps = 0  # metrica pedida no desafio: total de time steps do treino

    for taxi_run in range(epochs):
        state, _ = env.reset()
        done = False

        while not done:
            total_steps += 1
            if random.uniform(0, 1) < exploration:
                action = env.action_space.sample()  # explora
            else:
                action = np.argmax(q_table[state])  # usa o melhor Q conhecido

            next_state, reward, terminated, truncated, info = env.step(action)
            done = terminated or truncated

            prev_q = q_table[state, action]
            next_max_q = np.max(q_table[next_state])
            new_q = (1 - learning_rate) * prev_q + learning_rate * (reward + discount_factor * next_max_q)
            q_table[state, action] = new_q

            state = next_state

    return q_table, total_steps


def test_policy(env, q_table, n_trips=10):
    """Simula n_trips viagens usando a politica aprendida (sem sleep/clear_output),
    retornando a media de passos por viagem."""
    total_steps = 0
    for trip in range(n_trips):
        state, _ = env.reset()
        done = False
        steps = 0
        while not done:
            action = np.argmax(q_table[state])
            state, reward, terminated, truncated, info = env.step(action)
            done = terminated or truncated
            steps += 1
        total_steps += steps
    return total_steps / n_trips
```

```python
q_table, total_steps_treino = train_q_learning(streets, learning_rate=0.1, discount_factor=0.6, exploration=0.1, epochs=5000)
media_passos = test_policy(streets, q_table, n_trips=10)

print("Baseline (lr=0.1, discount=0.6, exploracao=0.1, epochs=5000)")
print("Total de passos no treino:", total_steps_treino)
print("Media de passos por viagem (10 viagens simuladas):", media_passos)
```

### Parte 2 - Experimentando hiperparametros

Testamos: baseline, poucas epocas (para ver quando o modelo comeca a "sofrer"), learning rate mais alto, discount factor mais alto e mais exploracao.

```python
configs = [
    ("baseline",           0.1, 0.6,  0.1, 5000),
    ("poucas_epochs",      0.1, 0.6,  0.1, 200),
    ("learning_rate_alto", 0.5, 0.6,  0.1, 5000),
    ("discount_alto",      0.1, 0.95, 0.1, 5000),
    ("mais_exploracao",    0.1, 0.6,  0.4, 5000),
]

print(f"{'config':18s} | lr    | discount | exploracao | epochs | passos treino | media passos/viagem")
for nome, lr, df, expl, ep in configs:
    qt, steps_train = train_q_learning(streets, lr, df, expl, ep)
    avg = test_policy(streets, qt, n_trips=10)
    print(f"{nome:18s} | {lr:<5} | {df:<8} | {expl:<10} | {ep:<6} | {steps_train:<13} | {avg:.1f}")
```

**Conclusao do Desafio 2:**

- **Poucas epocas (200):** o agente nao teve treino suficiente para aprender uma boa politica - a media de passos por viagem bateu no limite maximo do ambiente (200 passos), ou seja, o taxi frequentemente nem conseguia completar a corrida. Isso mostra visualmente "quando o modelo comeca a sofrer": abaixo de um certo numero de epocas, o Q-Learning nao converge.
- **Baseline (lr=0.1, discount=0.6, exploracao=0.1, epochs=5000):** com esses parametros "padrao" o modelo ainda nao converge de forma consistente dentro de 5000 epocas, ficando com uma media de passos bem mais alta que o ideal.
- **Learning rate mais alto (0.5):** convergiu para uma politica quase otima (media proxima de 13 passos por viagem, perto do otimo teorico do Taxi), usando menos passos totais de treino do que o baseline - ou seja, aprendeu mais rapido.
- **Discount factor mais alto (0.95):** tambem convergiu muito bem, pois o agente passou a valorizar mais as recompensas futuras (como o bonus de deixar o passageiro no lugar certo), nao so a recompensa imediata.
- **Mais exploracao (0.4):** surpreendentemente, foi a configuracao com o melhor resultado neste teste (media de ~13 passos). Isso mostra que, com o learning rate e discount factor padrao, uma exploracao maior ajudou o agente a conhecer mais estados diferentes dentro das mesmas 5000 epocas, evitando ficar preso em uma politica sub-otima aprendida cedo demais.

O experimento confirma o que foi visto na aula: nao e so o numero de epocas que importa, mas como cada epoca e aproveitada. Os parametros de learning rate, discount factor e taxa de exploracao interagem entre si, e o baseline "arbitrario" usado originalmente nao era necessariamente a melhor combinacao - varias configuracoes testadas aqui convergiram mais rapido e melhor do que ele.
