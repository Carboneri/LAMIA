```python
import re
# regex q separa cada campo de uma linha do log de acesso (host, data, request, status, referer, user agent etc)
format_pat = re.compile(
    r'(?P<host>[\d\.]+)\s'
    r'(?P<identity>\S*)\s'
    r'(?P<user>\S*)\s'
    r'\[(?P<time>.*?)\]\s'
    r'"(?P<request>.*?)"\s'
    r'(?P<status>\d+)\s'
    r'(?P<bytes>\S*)\s'
    r'"(?P<referer>.*?)"\s'
    r'"(?P<user_agent>.*?)"\s*'
)
```

```python
logPath = r"C:\Users\zinho\OneDrive\Documentos\Lamia\Card 11 - Prática Lidando com Dados do Mundo Real (II)\Aquivos_de_C%C3%B3digo\access_log.txt"  # caminho do log q vamos analisar
```

```python
URLCounts = {}  # dict p guardar qtas vezes cada URL aparece

with open(logPath, "r") as f:
    for line in (l.rstrip() for l in f):  # le o arquivo linha por linha
        match = format_pat.match(line)  # tenta casar a linha c o regex
        if match:  # so segue se achou os campos certinho
            access = match.groupdict()  # vira um dict
            request = access['request']  # pega so o campo da requisicao
            (action, URL, protocol) = request.split()  # separa em acao, URL e protocolo
            if URL in URLCounts:  # se ja apareceu antes, soma 1
                URLCounts[URL] = URLCounts[URL] + 1
            else:  # se nao, comeca contando do 1
                URLCounts[URL] = 1

results = sorted(URLCounts, key = lambda i: int(URLCounts[i]), reverse = True)  # ordena da URL mais vista p menos

for result in results[:20]:  # so as 20 mais vistas
    print(result + ": " + str(URLCounts[result]))
# da erro pq tem linha de log cuja requisicao nao tem exatamente 3 partes,
# ai o unpacking (action, URL, protocol) quebra pq falta valor p preencher as 3 variaveis
```
