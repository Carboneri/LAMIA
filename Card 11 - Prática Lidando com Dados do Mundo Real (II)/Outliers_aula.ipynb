```python
%matplotlib inline
import numpy as np

# gera 10000 valores de "renda" numa distribuicao normal (media 27000, desvio padrao 15000)
incomes = np.random.normal(27000, 15000, 10000)
incomes = np.append(incomes, [10000000000])  # add um outlier gigante de proposito (10 bilhoes)

import matplotlib.pyplot as plt
plt.hist(incomes, 50)  # histograma c 50 barras
plt.show()  # c o outlier o grafico fica espremido, quase nao da p ver a distribuicao normal
```

```python
incomes.mean()  # media fica toda distorcida pq o outlier gigante puxa ela pra cima
```

```python
def reject_outliers(data):
    u = np.median(data)  # usa mediana em vez de media pq mediana nao eh afetada pelo outlier
    s = np.std(data)      # desvio padrao dos dados
    # mantem so os valores dentro de 2 desvios padrao da mediana, descarta o resto (os outliers)
    filtered = [e for e in data if (u - 2 * s < e < u + 2 * s)]
    return filtered

filtered = reject_outliers(incomes)

plt.hist(filtered, 50)  # sem o outlier o histograma mostra certinho a distribuicao normal
plt.show()
```

```python
np.mean(filtered)  # media sem o outlier, bem mais proxima do valor esperado (~27000)
```
