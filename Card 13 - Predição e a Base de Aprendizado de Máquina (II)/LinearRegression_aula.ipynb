## Linear Regression

Vamos fabricar alguns dados que mostram uma relação aproximadamente linear entre a velocidade da página e o valor da compra:

```python
import numpy as np
import matplotlib.pyplot as plt

pageSpeeds = np.random.normal(3.0, 1, 1000)
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0, 0.1, 1000)) * 3

plt.scatter(pageSpeeds, purchaseAmount)
```

Como temos apenas duas variáveis (features), podemos simplificar e usar apenas o scipy.state.linregress:

```python
from scipy import stats

slope, intercept, r_value, p_value, std_err = stats.linregress(pageSpeeds, purchaseAmount)
```

Não surpreendentemente, nosso valor de R-quadrado mostra um ajuste muito bom:

```python
r_value ** 2
```

Vamos usar o slope e o intercept que obtivemos da regressão para plotar os valores previstos vs. observados:

```python
import matplotlib.pyplot as plt

def predict(x):
    return slope * x + intercept

fitLine = predict(pageSpeeds)

plt.scatter(pageSpeeds, purchaseAmount)
plt.plot(pageSpeeds, fitLine, c='r')
plt.show()
```

```python
pageSpeeds = np.random.normal(3.0, 1.0, 1000) ## nao mudei nada no eixo x 
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0, 2, 1000)) * 3 # aumentei o desvio padrao  e mantive a media em 0

plt.scatter(pageSpeeds, purchaseAmount) #como mudei o desvio padrao da formula para 2, ficou mais espalhado os pontos de quanto devem ser o preco por pagina.
```

```python
from scipy import stats
slope, intercept, r_value, p_value, std_err = stats.linregress(pageSpeeds, purchaseAmount)
```

```python
r_value ** 2 #o R-quadrado ficou horrivel por causa de estar muito espalhado 
```

```python
import matplotlib.pyplot as plt

def predict(x):
    return slope * x + intercept # formula pra prever

fitLine = predict(pageSpeeds)#aplica nos 1000 numeros gerados

plt.scatter(pageSpeeds, purchaseAmount)
plt.plot(pageSpeeds, fitLine, c='r') 
plt.show() #nao da pra prever nada e os pontos passam muito longe da reta 
```
